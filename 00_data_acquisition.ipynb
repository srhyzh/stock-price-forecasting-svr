```python
import pandas as pd
import numpy as np
import datetime
import yfinance as yf
```

```python
# Get the list of Dow Jones Industrial constituents from Wikipedia
url = 'https://en.wikipedia.org/wiki/Dow_Jones_Industrial_Average#Components'
tables = pd.read_html(url)
df = tables[1]
tickers = df['Symbol'].tolist()

# Set the start and end date for the data
start_date = '2002-01-01'
end_date = '2023-04-15'

# Get the historical data for each stock and store it in a dictionary
data = {}
for ticker in tickers:
    try:
        stock_data = yf.download(ticker, start=start_date, end=end_date)['Close']
        data[ticker] = stock_data
    except:
        print(f"Could not download data for {ticker}")

# Combine the data into a single dataframe
df = pd.DataFrame(data)

```
